Remove HTML Tag

```python
import re
def remove_html_tags(text):
    pattern=re.compile("<.*?>")
    return pattern.sub(r"",text)
```

```python
def remove_punctuation(text):
    for char in exclude:
        text=text.replace(char,"")
    return text
```

```python
stopwords.words("spanish")
```

```python
def remove_stopwords(text):
    new_text=[]
    for word in text.split():
        if word in stopwords.words("english"):
            new_text.append("")
        else:
            new_text.append(word)
    x=new_text[:]
    new_text.clear()
    return " ".join(x)
```

```python
remove_stopwords("'This is a sample text with some stopwords.'")
```

Handling Emojis

```python
import re

def remove_emojis(text):
    emoji_pattern = re.compile(
        "["
        "\U0001F600-\U0001F64F"
        "\U0001F300-\U0001F5FF"
        "\U0001F680-\U0001F6FF"
        "\U0001F700-\U0001F77F"
        "\U0001F780-\U0001F7FF"
        "\U0001F800-\U0001F8FF"
        "\U0001F900-\U0001F9FF"
        "\U0001FA00-\U0001FAFF"
        "\U00002702-\U000027B0"
        "\U000024C2-\U0001F251"
        "]+",
        flags=re.UNICODE
    )
    
    return emoji_pattern.sub(r'', text)
```

```python
print(remove_emojis("I love NLP ❤️😊. It is very interesting 🔥🚀!"))
```

Stemming

```python
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
sample="walk walks walking walked"
stem_words(sample)
```

Lemmatization


```python
import nltk
from nltk.stem import WordNetLemmatizer

wordnet_lemmatizer = WordNetLemmatizer()

text = "The children were running in the garden and studying their lessons. They studied hard and became better students."

punctuation = ".,!?:;"

sent_word = nltk.word_tokenize(text)

print("{0:20}{1:20}".format("Word", "Lemma"))

for word in sent_word:
    if word not in punctuation:
        print("{0:20}{1:20}".format(
            word,
            wordnet_lemmatizer.lemmatize(word,pos="v")
        ))
```
