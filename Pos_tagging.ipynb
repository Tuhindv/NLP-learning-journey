```python
import spacy
nlp=spacy.load("en_core_web_sm")
doc=nlp(u"I will google about facebook")
```

```python
doc.text
```

```python
doc[-1]
```

```python
doc[2].pos_
```

```python
doc[0].tag_
```

```python
spacy.explain("VB")
```

```python
for word in doc:
    print(word.text,"----->",word.pos_,word.tag_,spacy.explain(word.tag_))
```
