```python
sample_text = "<h2>Sample Heading</h2><p>The HTML <code class='w3-codespan'>&lt;sup&gt;</code> element defines superscript text.</p>"
```

```python
import re
def striphtml(data):
    p=re.compile(r'<.*?>')
    return p.sub("",data)
```

```python
striphtml(sample_text)
```

```python
import nltk
from nltk.tokenize import sent_tokenize,word_tokenize
```

```python
dummy="Artificial intelligence is one of the most important and fascinating technologies in the modern world. It allows computers to perform complex tasks that usually require human intelligence. Machine learning is a subfield of artificial intelligence that enables computers to learn from data and make predictions without being explicitly programmed. Natural language processing, also known as NLP, is an important area of artificial intelligence. It helps computers understand, interpret, and generate human language. NLP is used in many real-world applications such as chatbots, machine translation, sentiment analysis, speech recognition, text classification, and information retrieval."
```

```python
dummy
```

```python
sents = sent_tokenize(dummy)
sents
```

```python
for sent in sents:
    print(word_tokenize(sent))
```
